```python
def http_error(status):
    match status:
        case 400:
            return "bad request"

        case 404:
            return "not found"

        case 418:
            return "www.claw"

        case _:
            return "network down"    
                
```

```python
status_uptade=http_error(404)
print(status_uptade)
```

```python
def temp_checker(temp,unit):
    if unit=='f':
        return (temp-32)*(5/9)

    elif unit=='c':
        return temp*9/5+32


    else:
        return None    


print(temp_checker(50,'c'))
print(temp_checker(30,'f'))
```

```python
def rec(n):
    # base condition
    if n==0 or n==1:
        return 1

    return n*rec(n-1)


print(rec(6))  

```

```python
def enumerate(iterable, start=0):
    n = start
    for elem in iterable:
        yield n, elem
        n += 1
```
