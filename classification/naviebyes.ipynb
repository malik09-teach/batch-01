```python
# 1. OUR TRAINING DATA
training_data = [
    ("win free cash", "spam"),
    ("free cash prize", "spam"),
    ("lunch meeting tomorrow", "ham"),
    ("team meeting schedule", "ham")
]


```

```python
# 2. EXTRACT THE VOCABULARY AND COUNTS
spam_words = []
ham_words = []
spam_emails = 0
ham_emails = 0

for text, label in training_data:
    words = text.split()
    if label == "spam":
        spam_words.extend(words)
        spam_emails += 1
    else:
        ham_words.extend(words)
        ham_emails += 1

total_emails = spam_emails + ham_emails
vocab = set(spam_words + ham_words) # All unique words in the dataset
vocab_size = len(vocab)
```

```python
print(spam_words)
print(len(spam_words))
print(ham_words)
print(len(ham_words))
print(spam_emails)
print(ham_emails)
print(total_emails)
print(vocab)
print(vocab_size)
```

```python
# ==========================================
# THE BAYES THEOREM PREDICTION ALGORITHM
# ==========================================

def predict_naive_bayes(new_email):
    words_in_email = new_email.split()
    
    # STEP 1: Calculate P(C) - The Prior Probabilities
    p_spam = spam_emails / total_emails
    p_ham = ham_emails / total_emails
    
    # STEP 2 & 3: Calculate P(X|C) * P(C)
    # We start our final score with the Prior P(C)
    spam_score = p_spam
    ham_score = p_ham
    
    # Multiply by the Likelihood P(X|C) for every word in the new email
    for word in words_in_email:
        
        # Laplace Smoothing: (count of word in class + 1) / (total words in class + vocab size)
        spam_word_count = spam_words.count(word)
        p_word_given_spam = (spam_word_count + 1) / (len(spam_words) + vocab_size)
        
        ham_word_count = ham_words.count(word)
        p_word_given_ham = (ham_word_count + 1) / (len(ham_words) + vocab_size)
        
        # Multiply the probabilities ( P(X|C) * P(C) )
        spam_score = spam_score * p_word_given_spam
        ham_score = ham_score * p_word_given_ham

    # STEP 4: Ignore P(X) denominator (handled implicitly by just comparing numerators)
    
    # STEP 5: Compare and Predict
    print(f"Message: '{new_email}'")
    print(f"-> Math Score for SPAM: {spam_score:.6f}")
    print(f"-> Math Score for HAM:  {ham_score:.6f}")
    
    if spam_score > ham_score:
        print("PREDICTION: SPAM\n")
    else:
        print("PREDICTION: HAM\n")
```

```python


# --- Let's test it! ---
predict_naive_bayes("free cash")
predict_naive_bayes("meeting tomorrow")
predict_naive_bayes("free meeting")
```
