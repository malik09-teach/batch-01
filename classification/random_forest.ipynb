```python
import numpy as np 
import pandas as pd
from sklearn.datasets import make_classification
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import  accuracy_score
from sklearn.model_selection import  train_test_split
```

```python
x,y = make_classification(n_features=5, n_redundant=0, n_informative=5,n_clusters_per_class=1)
```

```python
x=pd.DataFrame(x, columns=[f'feature_{i}' for i in range(x.shape[1])])
y=pd.DataFrame(y,columns=["target" in y.shape])
```

```python
rf_model = RandomForestClassifier(
    n_estimators=5,        
    random_state=42
)


rf_model.fit(x,y)








'''rf_model = RandomForestClassifier(
    n_estimators=100,         # Builds 100 trees (B = 100)
    bootstrap=True,           # Enables Row Sampling (with replacement)
    max_samples=None,         # Each bootstrap sample matches the original dataset size
    max_features='sqrt',      # Enables Column Sampling (uses square root of total columns at each split)
    random_state=42
)'''
```

```python
predictions = rf_model.predict(x)
print(f"Random Forest Overall Accuracy: {accuracy_score(y, predictions):.2f}")
```
