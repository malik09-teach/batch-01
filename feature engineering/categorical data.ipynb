```python
import pandas as pd
from sklearn.compose import  ColumnTransformer
from sklearn.preprocessing import  OneHotEncoder
import seaborn as sns 

```

```python
data=sns.load_dataset('titanic')
```

```python
data.head()
```

```python
data.shape
```

```python
data['embark_town'].value_counts().plot(kind='pie',autopct='%%.2.f%%')
```

```python
ct=ColumnTransformer(transformers=[('cat',OneHotEncoder(sparse_output=False),['embark_town'])],remainder='passthrough')
```

```python
datan=ct.fit_transform(data)
```

```python
data.head()
```

```python
datan.head()
```

```python
df = pd.DataFrame(datan, columns=ct.get_feature_names_out())
df.head()
```
