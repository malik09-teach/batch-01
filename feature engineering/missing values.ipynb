# Missing Value Handling
This notebook shows how to download a dataset, inspect missing values, and handle them for both numeric and categorical columns.

```python
import seaborn as sns 
import pandas as pd 
import numpy as np
from sklearn.impute import  SimpleImputer
```

```python
data=sns.load_dataset('titanic')
```

```python
data.sample(5)
```

```python
data.shape
```

```python
data.info()
```

```python
data.isnull().sum()
```

```python
print(data.iloc[:,3:4])
```

```python
impute=SimpleImputer(missing_values=np.nan,strategy='mean')
```

```python
data.iloc[:,3:4]=impute.fit_transform(data.iloc[:,3:4])
```

```python
data.info()
```

```python
data.isna().sum()
```

```python
data['embark_town'].value_counts().plot(kind='pie',autopct='%.2f%%')
```

```python
impute=SimpleImputer(missing_values=np.nan,strategy='most_frequent')



```

```python
data.iloc[:,12::13]=impute.fit_transform(data.iloc[:,12::13])
```
